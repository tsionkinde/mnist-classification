```python
import os
import numpy as np
from sklearn.datasets import fetch_openml
from sklearn.metrics import accuracy_score

if not os.path.exists("mnist.npz"):
    mnist = fetch_openml("mnist_784", version=1, as_frame=False)
    X = mnist.data
    y = mnist.target
    np.savez_compressed("mnist.npz", X=X, y=y, allow_pickle=True)
    print("Dataset downloaded and saved.")
else:
    print("Dataset already exists.")
```

```python
# Load the saved file
with np.load("mnist.npz", allow_pickle=True) as data:
    X = data["X"]
    y = data["y"]

# Split into training (first 60,000) and testing (last 10,000)
X_train, X_test = X[:60000], X[60000:]
y_train, y_test = y[:60000], y[60000:]

# Scale the pixel values so they are between 0 and 1
X_train, X_test = X_train / 255.0, X_test / 255.0
print(f"Training data shape: {X_train.shape}")
print(f"Testing data shape: {X_test.shape}")

print("Data loaded, split, and scaled successfully!")
```

```python
from sklearn.linear_model import SGDClassifier

sgd_clf = SGDClassifier(random_state=42)
sgd_clf.fit(X_train, y_train)

print("Model training is finished!")
```

```python

from sklearn.metrics import accuracy_score
y_pred = sgd_clf.predict(X_test)
print("Accuracy score:", accuracy_score(y_test, y_pred))
y_pred = sgd_clf.predict(X_test)

# Calculate and print the accuracy score
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Test Accuracy: {accuracy * 100:.2f}%")
import sys
print(sys.version)
print(sys.executable)

```
